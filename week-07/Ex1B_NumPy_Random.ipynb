## Generating Random Choices from an Array

```python
students = np.array(["Bob","John","Mike","Sally","Larry","Asta","Noel","Lucy","Juvia","Erza","Gray","Natsu"])
```

```python
call_on = np.random.choice(students)
print(call_on)
```

```python
breakout_groups = np.random.choice(students, size=(3,4))
print(breakout_groups)

#This is not a good way to assign breakout groups because some of the names repeat.
#One person cannot be in multiple groups, so this would cause a huge issue. 
#Furthermore, since some of the names are repeated, that takes spots away from other students.
#For exapmle, when I ran this code the first time, "Gray" showed up twice, but "Juvia" is not in the breakout groups at all.
```

```python
call_on_weighted = np.random.choice(
    students,
    p=[.3,0,.05,.05,.1,.1,.1,.05,.1,.05,.05,.05], size=(5))
print(call_on_weighted)
```

## Generating a Normal Distribution

```python
bell_curve = np.random.normal(loc=10, scale=5, size=(4,5))
print(bell_curve)
```
