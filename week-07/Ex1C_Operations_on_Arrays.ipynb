## Permutations

```python
print(r_4)
```

```python
np.random.shuffle(r_4)
print(r_4)
```

### The first row is the same, but the rows below it are different

```python
new_shuffle = np.random.permutation(r_4)
print(new_shuffle)
print(r_4)
```

### I would use permutation if I needed to compare the original array to the new array.
### I would use shuffle if I needed to change the sequence of an array and its original sequence does not matter.

## Grocery Problem

```python
bag1 = np.array([2,3,4,0])
bag2= np.array([4,0,1,2])
```

```python
sum_bag1 = np.sum(bag1)
sum_bag2 = np.sum(bag2)

print(f"The first bag contains {sum_bag1} items, and the second bag contains {sum_bag2} items.")
```

```python
bags = np.add(bag1, bag2)
print(bags)
```

```python
print(f"There are {bags[0]} vegetables, {bags[1]} fruits, {bags[2]} boxes of snacks, and {bags[3]} beverages in the two grocery bags.")
```

## Teamwork

```python
ali = np.array([5,4,6,4,2])
ben = np.array([np.nan,5,5,3,4])
carla = np.array([3,6,np.nan,3,4])
```

```python
all_knocks = np.array([ali, ben, carla])
print(all_knocks)
```

```python
#Total number of doors each person knocked on for the week
print(np.sum(all_knocks, axis = 1))
```

```python
#Total number of doors the group collectively knocked on
print(np.sum(all_knocks))
```

```python
daily_totals = np.sum(all_knocks)
rolling_knocks = np.cumsum(daily_totals)
rolling_votes = rolling_knocks * 0.5
```

```python
print(daily_totals)
print(rolling_knocks)
print(rolling_votes)
```
