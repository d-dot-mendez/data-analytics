#  Exercise 3.A Getting familiar with Matplotlib

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Creating Basic Plots. 

```python
x=[1,2,3,4,5]
y=[2,3,5,7,11]
```

```python
plt.plot(x,y)
plt.show()
```

```python
#Updating plot with markers
plt.plot(x,y, marker='o')
plt.show()
```

```python
#Updating plot with labels
plt.plot(x,y, marker='o')
plt.title("Example Graph")
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.show()
```

## Styling Plots

```python
#Changing the line to a dash and to the color red
plt.plot(x,y, ls='--', c= 'r', marker='o')
plt.title("Example Graph")
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.show()
```

```python
#changing the merker style to green, size 10, and a square
plt.plot(x,y, ls='--', c= 'r', marker='s', ms=10, mfc='g')
plt.title("Example Graph")
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.show()
```

```python
#Experimenation 1
plt.plot(x,y, ls=':', c= 'r', marker='d', ms=10, mfc='m')
plt.title("Example Graph")
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.show()
```

```python
#Experimentation 2
plt.plot(x,y, ls=':', c= 'g', marker='*', ms=10, mec='b', mfc='c')
plt.title("Example Graph")
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.show()
```
