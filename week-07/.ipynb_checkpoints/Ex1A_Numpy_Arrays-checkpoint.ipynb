# Exercise 1.A Numpy: Manipulating Arrays

```python
import numpy as np
```

## Looop through Arrays

```python
arrA= np.array([[11,12,13], [14,15,16]])
print(arrA)
```

### First dimesion multiplied by 2

```python
for x in arrA:
    print(x*2)
    
```

### Second dimension multiplied by 2

```python
for x in arrA:
    for y in x:
        print(y*2)
```

### The difference between the first loop and the second loop is how they are displayed. The first loop is still displayed in brackets
### where as the second loop is displayed vertically. 

```python
for x in arrA:
    for y in x:
        loop_result = (x*2)
        print(loop_result)
#This result si different because the value that is being multiplied in this nested loop is the x.
    
```

## Joining and Stacking Arrays

```python
joined_array_1 = np.concatenate((arrA, arrA), axis=0)
print(joined_array_1)
```

```python
joined_array_2 = np.concatenate((arrA, arrA), axis=1)
print(joined_array_2)
```

```python
stacked_array_1 = np.stack((joined_array_1, joined_array_1))
print(stacked_array_1)
```

```python
stacked_array_2 = np.stack((joined_array_1, joined_array_2))
print(stacked_array_2)
#I received a value error becauyse the two arrays have a different shape.
#The first array is 2D with a shape of (4,3)
#The second array is 2D with a shape of (2,6)
```

```python
another_joined_array = np.concatenate((joined_array_1, arrA))
print(another_joined_array)
```

```python
another_2 = np.concatenate((joined_array_1, arrA), axis=1)
#This does not work because both arrays have different number of attributes on axis 0
```

```python
arrA_v = np.vstack((arrA, arrA))
print(arrA_v)

#ArrA is stacked on top of ArrA 
```

```python
arrA_h = np.hstack((arrA, arrA))
print(arrA_h)

#ArrA is stacked next to ArrA
```

```python
arrA_d = np.dstack((arrA, arrA))
print(arrA_d)

#ArrA is stacked in a way where it is layered on top of ArrA
```

## Splitting Arrays

```python
splitsky = np.array_split( joined_array_1, 2)
print(splitsky)

#There are two arrays, and each one is 2D
```

```python
spliterino = np.array_split( joined_array_1, 3)
print(spliterino)

#There are three arrays and each one is 2D
```

## Searching And Replacing Array Values

```python
result_1 = np.where(arrA%2 == 0, arrA,'odd')
print(result_1)
```

## Sorting and Filtering Arrays

```python
arrB = np.array([(7,6), (13,5), (7,3)])
arrB_sorted = np.sort(arrB)
print(arrB_sorted)
```

```python
print(arrB)
z = arrB < 7
arrC = arrB[z]
print(arrC)
```

```python
print(arrB)
y = arrB >= 7
arrD = arrB[y]
print(arrD)
```
