## Working with defining custom functions

```python
def favorite_things(name, movie, music):
    message = (f"Hi {name}!\n"
               f"Your favorite movie is {movie}.\n"
               f"Your favorite music is {music}.")
    return message
print(favorite_things("Denae", "Coraline", "Aqours"))
    
```

```python
def why_im_here(purpose):
    message = (f"This is my purpose. {purpose}")
    return message

print(why_im_here("I enjoy the Year Up United Data Analyst Program because /n"
                  "I now have an opportunity to create a better future for myself."))
```

```python
def favorite_place(place, why):
    message =(f"My favorite place is {place} because {why}.")
    return message
    
```

```python
print(favorite_place("Sakura's", "I like to east sushi"))
```

```python
print(favorite_place("Mochinut", "I like to east mochi donuts"))
```
