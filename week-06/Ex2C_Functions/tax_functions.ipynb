# Tax Rates
## This lesson teaches the student how to create functions using taxes 

```python
def get_soc_sec_tax(price):
    return price * 6.2
```

```python
def get_medicare_tax(price):
    return price *1.45
```

```python
def get_federal_tax(price, code):
    if code == 0:
        return price *.23
    elif code == 1:
        return price *.21
    elif code == 2:
        return price *.195
    elif code == 3: 
        return price*.185
    else:
        return price *.18
```

```python
#person 1, gross pay:750, code:0

print(get_soc_sec_tax(750))
print(get_medicare_tax(750))
print(get_federal_tax(750, 0))
```

```python
#person 2, gross pay:1550 code:2
print(get_soc_sec_tax(1550))
print(get_medicare_tax(1550))
print(get_federal_tax(1550, 2))
```

```python
#person 3, gross pay:1100 code:5
print(get_soc_sec_tax(1100))
print(get_medicare_tax(1100))
print(get_federal_tax(1100, 5))
```
