## Array Basics

```python
fruits = ["apple","banana", "cherry", "date"]
```

```python
arr_1 = np.array(fruits)
print(arr_1)
```

```python
print(arr_1.dtype)
#The dtypee is U6
#The first letter means the data type is an unsigned integer.
#The number represents the unicode character stored or the bytes per item
```

```python
arr_2 = np.array([[10,11,12],[13,14,15]])
print(arr_2)
```

```python
print(arr_2.T)
```

```python
#This did not chane the values stored in arr_2. The values are the same.
#It did, however, change the positioning of those values in arr_2. 
```

```python
arr_3 = np.array([[1,3,5],[2,4,6]])
print(arr_3)
```

```python
arr_4 = np.array([arr_3[0,2],arr_3[1,1]])
print(arr_4)
```

```python
arr_5 = np.array([
    [[1,1], [1,1]],
    [[1,1],[1,1]]
    ])
print(arr_5)
```
