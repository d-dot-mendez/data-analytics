```python
import datetime
```

# Today

```python
today = datetime.datetime.now() 
```

```python
day      = today.strftime("%A")
month    =  today.strftime("%B")
day_date =  today.strftime("%d")
year     = today.strftime("%Y")
```

```python
print(f"Today is{day},{month},{day_date},{year}")
```

# My birthday

```python
dob = datetime.datetime(1999, 8, 17)
birth_month = dob.strftime("%B")
birth_date  = dob.strftime("%d")
birth_year  = dob.strftime("%Y")
```

```python
print(f"My birthday is {dob}")
print(f"My birthday is {birth_month}, {birth_date}, {birth_year}")
```

# 90 days

```python
ninety_d = today + datetime.timedelta(days=90)
```

```python
print(ninety_d) #test
```

```python
n_date = ninety_d.strftime("%d")
n_month = ninety_d.strftime("%B")
n_year = ninety_d.strftime("%Y")
```

```python
print(f"Ninety days from today is {n_month}, {n_date}, {n_year}")
```

# Time for Dinner

```python
dinner_time = datetime.time(18, 30)
```

```python
hour = dinner_time.strftime("%I")
minutes = dinner_time.strftime("%M")
```

```python
print(f"Let's meet for dinner at {hour}:{minutes} PM")
```
