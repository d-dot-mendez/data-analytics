## Lab 1. Working with names- parsing strings


```python
full_name = "Michael Jackson" 
```

```python
def parse_and_display(name):
    for i in range(len(name)):
        if name[i] == ' ':
            first = name[:i]
            last = name[i+1:]
            print(f"Full name: {name}")
            print(f"First name: {first}")
            print(f"Last name: {last}")
```

```python
parse_and_display(full_name)
```

```python
test_list = ['Beyonce Knowles', 'Doja Cat', 'Ariana Grande']
```

```python
parse_and_display(test_list[0]) #Output for Beyonce
```

```python
parse_and_display(test_list[1]) #Output for Doja Cat
```

```python
parse_and_display(test_list[2]) #Output for Ariana Grande
```

```python
parse_and_display('Denae Mendez')
```
