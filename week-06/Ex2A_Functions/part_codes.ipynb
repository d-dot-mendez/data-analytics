```python
def get_supplier_code(part_code):
    colon = part_code.find(":")
    return part_code[:colon]

print(get_supplier_code("ACME:123-L"))
```

```python
def get_product_num(part_code):
    colon = part_code.find(":")
    dash = part_code.find("-")
    return part_code[colon + 1:dash]

print(get_product_num("ACME:123-L"))     # 123
print(get_product_num("DI:12345-M"))     # 12345
```

```python
def get_size(part_code):
    dash = part_code.find("-")
    return part_code[dash + 1:]

print(get_size("ACME:123-L"))            # L
print(get_size("ACE:1-12"))              # 12
```
